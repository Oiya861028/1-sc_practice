generate MA plots for different pre and post filter settings

```python
import pandas as pd
from pathlib import Path    

import matplotlib.pyplot as plt
# Custom MA_plot function
import sys
sys.path.append('/project/imoskowitz/yubin/1-sc_practice/Differential_Gene_Analysis/Utility/')
import MA_plot  # Now you can use MA_plot.custom_MA_plot() or similar
```

```python
pre_filters_path= Path("/project/imoskowitz/yubin/1-sc_practice/Differential_Gene_Analysis/DataFrames/Pre-pseudobulk-filter")
post_filters_path= Path("/project/imoskowitz/yubin/1-sc_practice/Differential_Gene_Analysis/DataFrames/Post-pseudobulk-filter")
pre_filters = []
for file in sorted(pre_filters_path.iterdir()):
    pre_filters.append(pd.read_csv(file))
    
post_filters = []
for file in sorted(post_filters_path.iterdir()):
    post_filters.append(pd.read_csv(file))
```

```python
pre_filters[0]
```

```python
fig, axs = plt.subplots(nrows = 3, ncols= 2, figsize=(8, 9), layout = "constrained")
axs_flatten = axs.flatten()
for i, filter in enumerate(pre_filters):
    MA_plot.custom_MA_plot(filter, axs_flatten[i], title = "MA Plot of " + str(i) +" cell filter")
fig.savefig("/project/imoskowitz/yubin/1-sc_practice/Differential_Gene_Analysis/Figures/MA_plot_pre_filter.svg")
```

```python
fig, axs = plt.subplots(nrows = 4, ncols= 2, figsize=(8, 12), layout = "constrained")
axs_flatten = axs.flatten()
for i, filter in enumerate(post_filters):
    MA_plot.custom_MA_plot(filter, axs_flatten[i], title = "MA Plot of " + str(i) +" sample filter")

fig.savefig("/project/imoskowitz/yubin/1-sc_practice/Differential_Gene_Analysis/Figures/MA_plot_post_filter.svg")
```
